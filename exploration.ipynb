```python
G = ox.graph_from_place(
    "Paris, France",
    network_type="drive"
)

orig = list(G.nodes)[0]
dest = list(G.nodes)[100]

dist, prev = dijkstra_osmnx(G, orig, dest)

path = reconstruct_path(prev, orig, dest)

print("Distance (m):", dist[dest])
print("Nombre de nodes:", len(path))
```

```python
ox.plot_graph_route(
    G,
    path,
    route_color="red",
    route_linewidth=4,
    node_size=0
)
```

```python
ox.save_graphml(G,"data/paris.graphml")
```
